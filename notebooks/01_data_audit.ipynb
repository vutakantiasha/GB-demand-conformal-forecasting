# 01 — Data Audit
Thin notebook for inspecting the package-generated master dataset. Core preprocessing lives in `src/`, not in this notebook.


```python
from pathlib import Path
import pandas as pd
ROOT = Path("..").resolve()
master = pd.read_csv(ROOT / "data/processed/master_dataset.csv.gz", parse_dates=["settlement_date"])
master.shape, master.head()
```

```python
master.groupby("split").agg(rows=("national_demand_mw","size"), days=("settlement_date","nunique"))
```

```python
master.groupby("settlement_date")["settlement_period"].nunique().value_counts().sort_index()
```

```python
cols=[c for c in ["demand_lag_2d_mw","demand_lag_7d_mw","embedded_wind_forecast_mw","embedded_solar_forecast_mw","neso_day_ahead_forecast_mw"] if c in master]
master[cols].isna().mean().sort_values(ascending=False)
```
