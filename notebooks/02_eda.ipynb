# 02 — Exploratory Analysis
Descriptive analysis only. Do not use the locked test set to select features or hyperparameters.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
ROOT=Path("..").resolve()
df=pd.read_csv(ROOT/"data/processed/master_dataset.csv.gz", parse_dates=["settlement_date"])
train=df[df.split=="training"].copy()
```

```python
profile=train.groupby("settlement_period")["national_demand_mw"].agg(["mean","std"])
profile.plot(y="mean", figsize=(10,4), title="Training-period mean national demand by settlement period")
plt.ylabel("MW"); plt.show()
```
